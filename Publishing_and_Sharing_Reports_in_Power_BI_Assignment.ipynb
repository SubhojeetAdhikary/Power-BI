## Easy Level


### Question 1
**Explain why publishing and sharing reports is considered more important than just building dashboards in Power BI Desktop.**

### Solution
Building a report in Power BI Desktop is only the development stage. Publishing makes the report available in Power BI Service, where business users can access it through a browser or mobile device.

Sharing adds collaboration, controlled access, scheduled refresh, centralized distribution, and governance.

Therefore, a useful BI solution is not only a well-designed dashboard; it must also deliver the right information securely to the right users.

**Power BI Workflow:**  
Power BI Desktop → Home → Publish → Select Workspace → Manage Access in Power BI Service → Share Report or Publish App

**Reporting Lifecycle:**  
Data Source → Power BI Desktop → Publish → Power BI Service → End Users


### Question 3
**What does publishing a report mean in Power BI? List any two outcomes that occur after a report is published.**

### Solution
Publishing means uploading a Power BI Desktop report to a selected workspace in Power BI Service.

Two important outcomes are:

1. The report becomes available in the Power BI Service workspace.
2. The semantic model is also made available in Power BI Service so the report can be refreshed and managed.

The published report can then be shared with users or distributed through a Power BI App.

### Solution

### Question 6
**Explain the difference between the following three sharing methods in Power BI Service: Direct Report Sharing, Workspace Access, and Power BI Apps. Also mention one use case for each.**

### Solution

**1. Direct Report Sharing**  
A specific report is shared directly with selected users.

**Use Case:** Sharing one sales report with a small group of managers.

**2. Workspace Access**  
Users are added to a workspace with a specific role such as Admin, Member, Contributor, or Viewer.

**Use Case:** A BI development team collaborating on reports.

**3. Power BI Apps**  
A collection of reports and dashboards is packaged and distributed to a larger audience.

**Use Case:** Sharing approved executive reports with an entire department.


### Question 7
**Describe the four workspace roles available in Power BI and explain how access control is managed using these roles.**

### Solution

**Admin**
- Full workspace control
- Can manage users and workspace settings

**Member**
- Can create, edit, and manage content
- Has more permissions than Contributor

**Contributor**
- Can create, edit, and delete workspace content
- Has limited permission-management capabilities

**Viewer**
- Can only view and interact with reports
- Cannot edit workspace content

Access should be given using the **least-privilege principle**, which means users receive only the permissions they actually need.

### Question 9
**Explain Row-Level Security (RLS) in Power BI. How does RLS ensure that different users see different data using the same report?**

### Solution
Row-Level Security (RLS) restricts the rows of data that a user can see while allowing multiple users to use the same report.

For example, a user responsible for the North Region can be allowed to view only North Region data.

**Example RLS DAX Filter:**


```python
[Region] = "North"
```

**Steps to Implement RLS in Power BI:**

1. Open Power BI Desktop.
2. Go to **Modeling → Manage Roles**.
3. Create a new role.
4. Apply a DAX filter to the required table.
5. Use **View As** to test the role.
6. Publish the report.
7. Open Power BI Service.
8. Go to the semantic model security settings.
9. Assign users or groups to the appropriate RLS role.

As a result, different users can open the same report but see different rows of data.


### Question 10
**Assume an organization shares reports without using Apps, workspace roles, or RLS. Discuss any three risks or issues that may arise related to security, governance, or data misuse, and explain how Power BI best practices help prevent them.**

### Solution

**1. Unauthorized Data Exposure**  
Users may see confidential or irrelevant data.

**Prevention:**  
Use Row-Level Security (RLS) and controlled permissions.

---

**2. Weak Governance and Uncontrolled Sharing**  
Sharing reports individually with many users becomes difficult to manage.

**Prevention:**  
Use Power BI Apps, security groups, and governed workspaces.

---

**3. Accidental Modification or Misuse**  
Giving too many users editing permissions can lead to accidental changes or deletion.

**Prevention:**  
Use workspace roles such as Admin, Member, Contributor, and Viewer based on the least-privilege principle.

---

A strong Power BI governance setup should:
- Separate developers from report consumers
- Use controlled workspace roles
- Use Power BI Apps for distribution
- Apply RLS when data access differs by user
- Regularly review user permissions
