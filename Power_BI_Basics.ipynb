## **Question 1: What is Power BI and why is it used in businesses?**

#### Ans

Power BI is a Business Intelligence and data visualization tool developed by Microsoft. It is used to connect, analyze, and visualize data in the form of reports and dashboards.

Businesses use Power BI to understand their data, identify trends, track performance, and make better business decisions.

## **Question 2: Name and explain the three main components of Power BI.**

#### Ans

The three main components of Power BI are:

- **Power BI Desktop**: It is used to connect data, clean and transform data, create data models, and design reports.

- **Power BI Service**: It is an online platform used to publish, share, and view Power BI reports and dashboards.

- **Power BI Mobile**: It is a mobile application that allows users to view and interact with Power BI reports and dashboards on mobile devices.

## **Question 3: Explain the Power BI workflow.**

#### Ans

The basic Power BI workflow is:

Connect Data → Clean Data → Create Data Model → Create Reports → Publish → Share

- First, connect Power BI to a data source.
- Clean and transform the data using Power Query.
- Create relationships and a data model.
- Create charts, tables, and other visualizations.
- Publish the report to Power BI Service.
- Share the report or dashboard with other users.

## **Question 5: Write step-by-step instructions to load the above dataset.**

#### Ans

The steps to load the Real Estate Dataset in Power BI are:

1. Open Power BI Desktop.
2. Click on Home → Get Data.
3. Select Excel Workbook.
4. Select the Real_Estate_Data_100_Rows.xlsx file.
5. Select the required sheet from the Navigator window.

7. Check the data and column types in Power Query.

## **Question 6: Define Data View, Report View, and Model View. Explain the purpose of each view.**

#### Ans

1. **Data View**:  Data View shows the actual data in rows and columns. It is used to check and understand the data.

2. **Report View**: Report View is used to create charts, tables, graphs, cards, and other visualizations.

3. **Model View**: Model View shows the tables and relationships between them. It is used to manage the data model and relationships.

## **Question 8: Split Owner Name to create two new columns as First Name and Last Name.**

#### Ans

To split the Owner Name column into First Name and Last Name in Power Query:

1. Open Power Query Editor.
2. Select the Owner Name column.
3. Go to Transform → Split Column → By Delimiter.
4. Select Space as the delimiter.
5. Select Left-most delimiter.
6. Click OK.
7. Two new columns will be created.
8. Rename the columns as First Name and Last Name.
9. Click Close & Apply.

Example:

If the Owner Name is Rahul Sharma:

- First Name → Rahul
- Last Name → Sharma
