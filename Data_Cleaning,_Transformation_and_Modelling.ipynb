# **Theoretical Questions**

## **Question 1: What is Data Loading in Power BI and why is it considered the first step of analysis?**

## Ans

Data Loading in Power BI means importing data from a source such as Excel, CSV, SQL Database, etc. into Power BI.

It is considered the first step because Power BI needs the data before we can clean, transform, visualize, and analyze it.

## **Question 2: Explain the difference between “Load” and “Transform Data” in Power BI.**

## Ans

**Load**: Loads the data directly into Power BI without making major changes.

**Transform Data**: Opens Power Query Editor, where we can clean and modify the data before loading it into Power BI.

For example, we can remove duplicates, replace null values, change data types, and create new columns using Power Query.

## **Question 3: What is a Fact Table and a Dimension Table? Give examples from the dataset.**

## Ans

**Fact Table**: A fact table contains measurable numerical data used for analysis.

Example from the dataset:

- Confirmed_Cases
- Recovered_Cases
- Deaths
- Active_Cases
- Tests_Conducted
- Hospitalized

**Dimension Table**: A dimension table contains descriptive information used to categorize or filter the data.

Examples from the dataset:

- Country
- State
- Date
- Vaccination_Status

## **Question 4: Why is Star Schema preferred over Snowflake Schema in Power BI?**

## Ans

Star Schema is preferred because it is simple and easy to understand. It usually has one central Fact Table connected to multiple Dimension Tables.

Advantages:

- Simple data model
- Better performance
- Easier to create relationships
- Easier for reporting and analysis
- Requires fewer joins

## **Question 7: Create a new column to calculate Recovery Rate.**

## Ans

Recovery Rate can be calculated using:

Recovery Rate = (Recovered Cases / Confirmed Cases) × 100

In Power BI, create a New Column using DAX:

Recovery Rate =

    DIVIDE('coronavirus_2020_dataset'[Recovered_Cases],
    'coronavirus_2020_dataset'[Confirmed_Cases],
    0) * 100

The new column shows the percentage of confirmed cases that have recovered.
